```python
import random

# Part A: Machine Initialization

machines = ['A', 'B', 'C']

N_play = 10
epsilon = 0.2

# Part B: Initialize N, R and Q tables

N = {'A': 0, 'B': 0, 'C': 0}

R = {'A': [], 'B': [], 'C': []}

Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

total_reward = 0

print("Modified Epsilon-Greedy Multi-Armed Bandit")
print("------------------------------------------")

# Part C: Epsilon-Greedy Decision

for play in range(1, N_play + 1):

    r = random.random()

    # Find machine with highest Q value
    best_machine = max(Q, key=Q.get)

    if r < epsilon:
        # Exploration: Select a machine except the best machine
        available = [m for m in machines if m != best_machine]
        selected = random.choice(available)

        decision = "Exploration"

    else:
        # Exploitation: Select machine with highest Q value
        selected = best_machine

        decision = "Exploitation"

    print("\nPlay:", play)
    print("Random number:", round(r, 3))
    print("Decision:", decision)
    print("Selected Machine:", selected)

    # Part D: Reward Collection

    reward = float(input("Enter reward: "))

    # Part E: Q-Value Update

    N[selected] += 1

    R[selected].append(reward)

    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    total_reward += reward

    print("Updated N:", N)
    print("Updated Q:", {
        m: round(Q[m], 2) for m in machines
    })


# Part F: Display Final Results

print("\n========== FINAL RESULTS ==========")

print("\nParticipation Count (N):")

for m in machines:
    print(f"Machine {m}: {N[m]}")

print("\nRewards (R):")

for m in machines:
    print(f"Machine {m}: {R[m]}")

print("\nEstimated Q Values:")

for m in machines:
    print(f"Q({m}) = {Q[m]:.2f}")

print("\nTotal Reward Received:", total_reward)

best_machine = max(Q, key=Q.get)

print("Greedy Best Machine:", best_machine)3
```
