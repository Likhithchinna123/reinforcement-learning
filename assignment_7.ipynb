```python
%%writefile grid_env_det.py

states = ["S0", "S1", "S3", "S4", "G"]
actions = ["up", "down", "left", "right"]

gamma = 0.9
goal_reward = 10
normal_reward = -1

transitions = {
    "S0": {"up": "S0", "down": "S0", "left": "S0", "right": "S1"},
    "S1": {"up": "S1", "down": "S3", "left": "S0", "right": "G"},
    "S3": {"up": "S1", "down": "S3", "left": "S3", "right": "S4"},
    "S4": {"up": "G", "down": "S4", "left": "S3", "right": "S4"},
    "G": {"up": "G", "down": "G", "left": "G", "right": "G"}
}

def get_next_state(state, action):
    return transitions[state][action]

def get_reward(next_state):
    if next_state == "G":
        return goal_reward
    else:
        return normal_reward
```

```python
from grid_env_det import *

print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)

state = input("Enter current state (S0, S1, S3, S4, G): ")
action = input("Enter action (up, down, left, right): ")

if state in states and action in actions:
    next_state = get_next_state(state, action)
    reward = get_reward(next_state)

    print("Current State:", state)
    print("Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward)
else:
    print("Invalid state or action")
```
