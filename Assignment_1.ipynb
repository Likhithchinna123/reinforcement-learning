```python
# Multi-Armed Bandit - Arithmetic Mean Method

# Input number of machines
n = int(input("Enter number of machines: "))

machines = {}

# Input rewards for each machine
for i in range(n):
    name = input(f"\nEnter machine name {i + 1}: ")
    rewards = list(map(
        float,
        input(f"Enter rewards for machine {name}: ").split()
    ))

    machines[name] = rewards

# Calculate Q values
q_values = {}

for name, rewards in machines.items():
    q = sum(rewards) / len(rewards)
    q_values[name] = q

    print(f"\nMachine {name}")
    print("Rewards:", rewards)
    print(f"Q({name}) = {q:.2f}")

# Select machine with highest Q value
best_machine = max(q_values, key=q_values.get)

print("\nGreedy Selected Machine:", best_machine)
```
