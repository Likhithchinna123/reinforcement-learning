```python
import math

# Part A: Machine Initialization

machines = ['A', 'B', 'C']

N_play = 10

# Part B: Initialize N, R, UCB and Q tables

N = {'A': 0, 'B': 0, 'C': 0}

R = {'A': [], 'B': [], 'C': []}

Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0}

total_reward = 0

print("Upper Confidence Bound - Multi Armed Bandit")
print("--------------------------------------------")

# Part C: UCB Calculation and Machine Selection

for play in range(1, N_play + 1):

    print("\nPlay:", play)

    # First select every machine at least once
    unplayed = [m for m in machines if N[m] == 0]

    if unplayed:
        selected = unplayed[0]

        print("Machine not played before:", selected)

    else:
        # Calculate UCB for every machine
        for m in machines:
            UCB[m] = Q[m] + math.sqrt(
                (2 * math.log(play)) / N[m]
            )

        print("\nUCB Values:")

        for m in machines:
            print(f"UCB({m}) = {UCB[m]:.4f}")

        # Select machine with highest UCB
        selected = max(UCB, key=UCB.get)

    print("\nSelected Machine:", selected)

    # Part D: Reward Collection

    reward = float(input("Enter reward: "))

    # Part E: Q-Value Update

    N[selected] += 1

    R[selected].append(reward)

    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    total_reward += reward

    print("Updated N:", N)

    print("Updated Q:", {
        m: round(Q[m], 2) for m in machines
    })


# Part F: Display Final Results

print("\n========== FINAL RESULTS ==========")

print("\nParticipation Count (N):")

for m in machines:
    print(f"Machine {m}: {N[m]}")

print("\nRewards (R):")

for m in machines:
    print(f"Machine {m}: {R[m]}")

# Calculate final UCB values
print("\nFinal UCB Values:")

for m in machines:
    if N[m] > 0:
        UCB[m] = Q[m] + math.sqrt(
            (2 * math.log(N_play)) / N[m]
        )
    else:
        UCB[m] = float('inf')

    print(f"UCB({m}) = {UCB[m]:.4f}")

print("\nEstimated Q Values:")

for m in machines:
    print(f"Q({m}) = {Q[m]:.2f}")

print("\nTotal Reward Received:", total_reward)

best_machine = max(UCB, key=UCB.get)

print("Best Machine according to Final UCB:", best_machine)
```
