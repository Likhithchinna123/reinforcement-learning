```python
# Greedy Algorithm using Arithmetic Mean

n = int(input("Enter number of machines: "))

machines = {}

for i in range(n):
    name = input("\nEnter machine name: ")
    rewards = list(map(int, input("Enter observed rewards: ").split()))

    machines[name] = rewards

# Calculate Q values
q_values = {}

for name, rewards in machines.items():
    q = sum(rewards) / len(rewards)
    q_values[name] = q

    print("\nMachine", name)
    print("Rewards:", rewards)
    print(f"Q({name}) = {q:.2f}")

# Select machine with highest Q value
best_machine = max(q_values, key=q_values.get)

print("\nGreedy Selected Machine:", best_machine)
print("Highest Q Value:", q_values[best_machine])
```
